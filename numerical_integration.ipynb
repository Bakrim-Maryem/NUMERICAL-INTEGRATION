```python
import math

def f(x):
    return math.exp(-x**2)

def trapez(f, a, b, n):
    h = (b - a) / n
    S = f(a) + f(b)
    for i in range(1, n):
        x = a + i * h
        S = S + 2 * f(x)
    I = h / 2 * S
    return I
```

```python
def simpson(f, a, b, n):
    h = (b - a) / n
    x0 = a
    I = f(x0)
    for i in range(1, n):
        xi = a + i * h
        if i % 2 == 1:
            I = I + 4 * f(xi)
        else:
            I = I + 2 * f(xi)
    xn = b
    I = I + f(xn)
    return  I * h / 3
```
